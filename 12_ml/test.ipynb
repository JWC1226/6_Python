#### Python 리스트에는 다양한 타입의 데이터를 담을 수 있다

```python
data_list = [1, "이", 3.0, [4]]
print(f"data_list : {data_list}")
print(f"각 요소 별 타입: {[type(x).__name__ for x in data_list]}")
data_list2 = [10, 20, 30, 40]
```

- Python의 리스트는 데이터를 구분하지 않고 담을 수 있는 점이 편리할 수 있지만, 다른 리스트와 연산을 하고자 할 때는 문제가 될 수 있다!
- 넘파이 배열 (ndarray)은 데이터 타입이 하나로 고정된 구조이다 > 연산이 쉽다
